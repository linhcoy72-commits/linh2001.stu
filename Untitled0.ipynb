```python
import csv      # Nhập hộp dụng cụ xử lý tệp CSV
import random   # Nhập hộp dụng cụ trích xuất ngẫu nhiên
import time     # Nhập hộp dụng cụ đồng hồ bấm giờ để đo thời gian
```

```python
filename = "my_words.csv"  # Tên tệp chúng ta sẽ đọc.
words = []                 # Tạo một 'giỏ trống (danh sách)' để chứa các từ vựng.

try:
    # Mở tệp (open) và đặt tên gọi thay thế là f. (Sử dụng utf-8-sig để tiếng Hàn không bị lỗi font)
    with open(filename, 'r', encoding='utf-8-sig') as f:
        # Sử dụng hộp dụng cụ csv để chuẩn bị đọc tệp từng dòng một.
        reader = csv.reader(f)

        # Lặp lại (for) đọc từng dòng từ đầu đến cuối tệp.
        for row in reader:
            # Nếu không phải là dòng trống (có từ 2 dữ liệu trở lên)
            if len(row) >= 2:
                # Gắn nhãn '한국어' (Tiếng Hàn) và '뜻' (Ý nghĩa) để tạo thành cặp (dictionary) và đưa vào giỏ words.
                # strip() đóng vai trò xóa khoảng trắng thừa ở đầu và cuối chữ.
                words.append({'한국어': row[0].strip(), '뜻': row[1].strip()})

    # Khi kết thúc việc đọc tệp, thông báo thành công cùng với tổng số từ vựng đã được đưa vào màn hình.
    print(f"🎉 Thành công! Đã tải tổng cộng {len(words)} từ vựng.")
except FileNotFoundError:
    # Nếu tên tệp sai hoặc chưa được tải lên, thay vì máy tính bị đứng, nó sẽ hiển thị thông báo lỗi này.
    print(f"⚠️ Lỗi: Không tìm thấy tệp '{filename}'. Vui lòng kiểm tra lại xem tệp đã được tải lên thư mục bên trái chưa nhé!")
```

```python
# Kiểm tra trước xem các từ đã vào giỏ words (danh sách) ở ô trên thành công chưa.
if len(words) > 0:
    while True: # Lặp lại vô hạn để đưa ra câu đố.

        # 1. Đưa ra câu hỏi
        # Sử dụng hộp dụng cụ random để chọn ngẫu nhiên một phần tử trong giỏ từ vựng.
        question = random.choice(words)
        kor_word = question['한국어'] # Phần ngoại ngữ/câu hỏi của từ được chọn
        meaning = question['뜻']    # Phần đáp án của từ được chọn

        # In câu hỏi và thời gian giới hạn lên màn hình một cách đẹp mắt.
        print(f"\n📝 Câu hỏi: Nghĩa của từ '{kor_word}' là gì? (⏳ Giới hạn thời gian 15 giây!)")

        # --------------------------------------------------------
        # 2. Bắt đầu đếm giờ! (Bật đồng hồ bấm giờ)
        # --------------------------------------------------------
        # Đo thời gian hiện tại ngay tại khoảnh khắc câu hỏi hiển thị trên màn hình và lưu vào biến start_time.
        start_time = time.time()

        # --------------------------------------------------------
        # 3. Nhận đầu vào từ người dùng
        # --------------------------------------------------------
        # Chờ người dùng nhập đáp án từ bàn phím và nhấn Enter.
        user_answer = input("Nhập đáp án của bạn (nhập '종료' để kết thúc): ")

        # --------------------------------------------------------
        # 4. Kết thúc đếm giờ! (Tắt đồng hồ bấm giờ và tính thời gian đã mất)
        # --------------------------------------------------------
        # Đo thời gian ngay sau khi người dùng nhấn Enter và lưu vào end_time.
        end_time = time.time()

        # Tính khoảng thời gian giữa hai thời điểm (thời gian kết thúc - thời gian bắt đầu) và lưu vào biến elapsed_time.
        elapsed_time = end_time - start_time

        # --------------------------------------------------------
        # 5. Kiểm tra đáp án và xử lý kết thúc
        # --------------------------------------------------------
        # Nếu người dùng nhập '종료', thoát khỏi vòng lặp vô hạn while (break).
        if user_answer == '종료':
            print("Kết thúc trò chơi đố vui. Bạn đã vất vả rồi!")
            break

        # Kiểm tra xem giá trị được nhập có hoàn toàn trùng khớp với đáp án (==) hay không,
        # và thời gian mất đi có từ 15 giây trở xuống (<=) hay không đồng thời bằng ký hiệu 'and'.
        if user_answer == meaning and elapsed_time <= 15.0:
            print(f"✅ Chính xác! Tuyệt vời! (Thời gian thực hiện: {elapsed_time:.1f} giây)")
        elif user_answer == meaning:
            # Trường hợp trả lời đúng đáp án nhưng quá 15 giây.
            print(f"⏰ Hết giờ! Bạn đã trả lời đúng nhưng quá chậm. (Thời gian thực hiện: {elapsed_time:.1f} giây)")
        else:
            print(f"❌ Sai rồi. Đáp án đúng là '{meaning}'.")
else:
    print("Không có dữ liệu từ vựng. Vui lòng chạy ô Step 3 ở trên trước để tải từ vựng về.")
```

```python
import csv
import random
import time

# 1. Tải danh sách từ vựng từ tệp CSV
filename = "my_words.csv"
words = []

try:
    with open(filename, 'r', encoding='utf-8-sig') as f:
        reader = csv.reader(f)
        for row in reader:
            if len(row) >= 2:
                words.append({'한국어': row[0].strip(), '뜻': row[1].strip()})
    print(f"🎉 Thành công! Đã tải tổng cộng {len(words)} từ vựng.")
except FileNotFoundError:
    print(f"⚠️ Lỗi: Không tìm thấy tệp '{filename}'. Vui lòng kiểm tra lại!")

# 2. Bắt đầu trò chơi Quiz Bot (có tính điểm)
if len(words) > 0:
    score = 0  # Khởi tạo biến lưu tổng điểm

    while True:
        # Chọn ngẫu nhiên 1 câu hỏi
        question = random.choice(words)
        kor_word = question['한국어']
        meaning = question['뜻']

        print(f"\n📝 Câu hỏi: Nghĩa của từ '{kor_word}' là gì? (⏳ Giới hạn thời gian 15 giây!)")

        # Bắt đầu đếm giờ
        start_time = time.time()

        # Nhận đáp án từ người dùng
        user_answer = input("Nhập đáp án của bạn (nhập '종료' để kết thúc): ")

        # Kết thúc đếm giờ
        end_time = time.time()
        elapsed_time = end_time - start_time

        # Kiểm tra nếu người dùng muốn dừng trò chơi
        if user_answer == '종료':
            print("\n----------------------------------------")
            print(f"🏁 Kết thúc trò chơi đố vui! Tổng điểm cuối cùng của bạn là: {score} điểm.")
            print("Bạn đã vất vả rồi!")
            print("----------------------------------------")
            break

        # Xử lý kiểm tra đáp án & cộng điểm
        if user_answer == meaning and elapsed_time <= 15.0:
            score += 10  # Cộng 10 điểm khi đúng và trong thời gian
            print(f"✅ Chính xác! Tuyệt vời! (+10 điểm | Thời gian thực hiện: {elapsed_time:.1f} giây)")
            print(f"📊 Điểm hiện tại: {score} điểm")
        elif user_answer == meaning:
            print(f"⏰ Hết giờ! Bạn đã trả lời đúng nhưng quá chậm (+0 điểm | Thời gian thực hiện: {elapsed_time:.1f} giây)")
            print(f"📊 Điểm hiện tại: {score} điểm")
        else:
            print(f"❌ Sai rồi (+0 điểm). Đáp án đúng là '{meaning}'.")
            print(f"📊 Điểm hiện tại: {score} điểm")
else:
    print("Không có dữ liệu từ vựng. Vui lòng kiểm tra tệp my_words.csv.")
```
